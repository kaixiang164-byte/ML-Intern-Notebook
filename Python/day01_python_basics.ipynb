```python
## practice 1: find mean( version 1& 2 )
def mean_1(numbers):
  total = 0
  for x in numbers:
    total = total + x
  return total/ len(numbers)

mean_1(numbers)

def mean_2(numbers):
  total = 0
  count = 0
  for x in numbers:
    total= total + x
    count = count + 1
  return total/count

mean_2(numbers)
```

```python
## practice 2: square_mean
def smean(numbers):
  total = 0
  count = 0

  for x in numbers:
    total = total + x*x
    count = count + 1
  return total/count

smean([1,3])
```

```python
## practice 3: variance
def variance(numbers):
  total = 0
  mu = mean_1(numbers)

  for x in numbers:
    total = total + (x - mu)**2

  return total/len(numbers)

variance([1,3])
```

```python
## practice 4: standardization
import numpy as np

def standardize(numbers):

  mu = mean_1(numbers)
  sigma = (variance(numbers))**0.5
  standardization_list = []

  for x in numbers:
    standardization_list.append((x - mu)/sigma)

  return standardization_list

standardize([1,3])

```

```python
## leetcode two sum double loop
def two_sum_double_loop(nums,target):
  search_list = []

  for i in range(len(nums)):
    for j in range(i+1,len(nums)):
      if nums[i] + nums [j] == target:
        search_list.append(i)
        search_list.append(j)
        return search_list

  return search_list



two_sum_double_loop(nums=[2,7,11,15],target=17)

```

```python
## leetcode two sum dict

def two_sum_dict_loop(nums, target):
  seen = {}

  for pos,x in enumerate(nums):
    print(pos,x)
    print("current: ", x)
    print("need: ", target - x)
    print("seen: ", seen)

    if (target - x) in seen:
      return [seen[target - x], pos]

    seen[x] = pos

  return []

two_sum_dict_loop(nums = [2,7,11,2,15],target = 4)
```

```python
## counting frequency
words = ["apple","banana","apple","orange","banana","apple"]

def counting_fruits(words):
  fruit_dict = {}
  apple = 0
  banana = 0
  orange = 0
  for word in words:
    if word == "apple":
      apple = apple + 1
      fruit_dict[word] = apple
    elif word =="banana":
      banana = banana + 1
      fruit_dict[word] = banana
    else:
      orange = orange + 1
      fruit_dict[word] = orange
  return fruit_dict

counting_fruits(words)
```

```python
## counting frequency correct version
def counting_frequency(words):
  words_dict = {}
  for word in words:
    if word in words_dict:
      words_dict[word] += 1
    else:
      words_dict[word] = 1
  return words_dict

counting_frequency(["apple","banana","apple","orange","banana","apple"])
```

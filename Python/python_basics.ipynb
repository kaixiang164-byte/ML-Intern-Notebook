```python
## practice 1: find mean( version 1& 2 )
def mean_1(numbers):
  total = 0
  for x in numbers:
    total = total + x
  return total/ len(numbers)

mean_1(numbers)

def mean_2(numbers):
  total = 0
  count = 0
  for x in numbers:
    total= total + x
    count = count + 1
  return total/count

mean_2(numbers)
```

```python
## practice 2: square_mean
def smean(numbers):
  total = 0
  count = 0

  for x in numbers:
    total = total + x*x
    count = count + 1
  return total/count

smean([1,3])
```

```python
## practice 3: variance
def variance(numbers):
  total = 0
  mu = mean_1(numbers)

  for x in numbers:
    total = total + (x - mu)**2

  return total/len(numbers)

variance([1,3])
```

```python
## practice 4: standardization
import numpy as np

def standardize(numbers):

  mu = mean_1(numbers)
  sigma = (variance(numbers))**0.5
  standardization_list = []

  for x in numbers:
    standardization_list.append((x - mu)/sigma)

  return standardization_list

standardize([1,3])

```

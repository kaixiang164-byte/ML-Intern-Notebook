```python
import pandas as pd

data = {
    "hours":[1,2,3,4,5,6,7,8],
    "score":[50,55,65,70,75,80,90,95]
}

df = pd.DataFrame(data)

df.shape
```

```python
## Step 1: create X and y
X = df[["hours"]]
y = df["score"]
print(X.shape)
print(y.shape)
```

```python
## Step 2: test train split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.25, #25% of data is used for test
    random_state = 42
)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
## first model -- linear regression

# step 1: import model
from sklearn.linear_model import LinearRegression

# step 2:create model 
model = LinearRegression()

# step 3: training the model
model.fit(X_train, y_train)

# step 4: parameters f the model
print(model.intercept_)
print(model.coef_)
```

```python
# prediction
y_pred = model.predict(X_test)
print(y_pred)
print(y_test)
```

```python
## evaluation of the model
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_pred, y_test)
rmse = mse**0.5
r2 = r2_score(y_pred, y_test)

print("MSE: ", mse)
print("RMSE: ", rmse)
print("R2: ", r2)
print("The model explains approximately", r2 * 100, "% of the variance in the target variable.")
```

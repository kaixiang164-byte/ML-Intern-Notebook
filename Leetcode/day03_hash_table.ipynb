```python
### review two sum
def two_sum_brute(nums,target):

    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[j] == target -nums[i]:
                return [i,j]

    return []

two_sum_brute([2,8,11,7,15],9)
```

```python
def two_sum_hash(nums,target):
    seen = {}
    for pos, x in enumerate(nums):
        if target - x in seen:
            return[seen[target - x], pos]
        seen[x] = pos
    return []

two_sum_hash([2,8,4,4,11,7,15],8)
```

```python
### leetcode 217
## given nums = [1,2,3,1] return true
## given nums = [1,2,3,4] return false
def contains_duplicate(nums):
    seen = {}
    for x in nums:
        if x in seen:
            return True
        seen[x] = 1
    return False
contains_duplicate([1,1,2,2,3,4])
```

```python
## leetcode 217 recomended version
def contains_duplicate(nums):
    seen=set()

    for x in nums:
        if x in seen:
            return True
        seen.add(x)

    return False
```

```python
## Leetcode 242
# 遍历string
def is_anagram(s,t):
    count = {}
    for char_s in s:
        if char_s in count:
            count[char_s] += 1
        else:
            count[char_s] = 1
    for char_t in t:
        if char_t in count:
            count[char_t] -= 1
        else:
            return False
    for x in count:
        if count[x] == 0:
            return True
        else:
            return False

print(is_anagram("aabb","bbaa"))
print(is_anagram("cat","rat"))
print(is_anagram("anagram","nagaram"))
```

```python
## Leetcode 49
from tokenize import group


sorted("eat")
sorted("tea")

def groupAnagrams(strs):
    group = {}
    for word in strs:
        if tuple(sorted(word)) in group:
            group[tuple(sorted(word))].append(word)
        else:
            group[tuple(sorted(word))] = [word]
    return list(group.values())

groupAnagrams(["eat","tea","tan","ate","nat","bat"])
```

```python
## better version
def groupAnagrams(strs):

    group = {}

    for word in strs:

        key = tuple(sorted(word))

        if key in group:
            group[key].append(word)

        else:
            group[key] = [word]

    return list(group.values())
```

```python
# Complexity Analysis
#
# Time Complexity:
#
# Assume:
# - n = number of strings
# - k = average length of each string
#
# For each word:
# sorted(word) takes O(k log k)
# because sorting k characters requires k log k time.
#
# We process all n strings:
# Time Complexity = O(n * k log k)
#
#
# Space Complexity:
#
# We use a dictionary:
# key -> list of words
#
# The dictionary stores all grouped strings.
# Therefore:
# Space Complexity = O(nk)
#
#
# Key Insight:
#
# The main idea is to design a signature for each word:
#
# key = tuple(sorted(word))
#
# All anagrams have the same key,
# so we can group them efficiently using a hash map.
```

```python

nums = [0, 1, 0, 3, 12]
non_zero = []

for num in nums:
    if num != 0:
        non_zero.append(num)

print(non_zero)
```

```python
nums = [0, 1, 0, 3, 12]
next_pos = 0

nums[next_pos], nums[1] = nums[1], nums[next_pos]
next_pos += 1

print(nums)
print(next_pos)
```

```python
## leetcode 283. Move Zeroes
def move_non_zero(nums):
    next_pos = 0
    for num in range(len(nums)):
        if nums[num] != 0:
            nums[next_pos],nums[num] = nums[num],nums[next_pos]
            next_pos += 1
    return nums
print(move_non_zero([0,1,0,3,12]))
print(move_non_zero([0,0,1]))
```
